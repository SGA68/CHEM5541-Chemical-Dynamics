### _Pre-equilibrium:_

Consider

$$A+B \rightarrow C \rightarrow D$$

with $k_b^{'} = 0$, $k_a^{'} >> k_B \rightarrow$ intermediate is in "pre-equilibrium" w/reactants 

Pre-equilibrium constant:
$$K_{pre} = \frac{[C]}{[A][B]}$$

Product formation rate:

$$\frac{dD}{dt} = k_b C$$

$$\frac{dD}{dt} = k_b K_{pre} A B$$

looks like 2nd order reaction w/ effective rate constant

$$k_{eff} = k_b K_{pre} = k_b \frac{k_a}{k_a^{'}}$$

### _Parallel reactions_

$A$ can go into $B$ or $C$ with $k_b$ or $k_c$

Differential rate laws:

$$\frac{dA}{dt} = -k_b A - k_c A = -(k_b+k_c)A$$

$$\frac{dB}{dt} = k_b A$$

$$\frac{dC}{dt} = k_c A$$

Reactants:

$$\int_{A_0}^{A} \frac{dA}{A} = - \int_{0}^{t} (k_b+k_c) dt$$

$$A(t) = A_0 e^{-(k_b+k_c)t}$$

Products:

$$\frac{dB}{dt} = k_b A_0 e^{-(k_b+k_c)t}$$

$$\int_{B_0}^{B} dB^{'} = k_b A\int_{0}^{t} e^{-(k_b+k_c)t^{'}} dt^{'}$$

$$B(t) = - \frac{k_b A_0}{k_b + k_c} (e^{-(k_b+k_c)t^{'}})|_0^t = - \frac{k_b A_0}{k_b + k_c} (e^{-(k_b+k_c)t} - 1) =\frac{k_b A_0}{k_b + k_c} (1 - e^{-(k_b+k_c)t})$$

$$C(t) = \frac{k_c A_0}{k_b+k_c} (1-e^{-(k_b+k_c)})$$

This can be generalized to arbitrary numnber of pathways:

$$A = A_0 e^{-\sum_j k_j t}$$

$$[j] = [A_0] \frac{k_j}{\sum_n k_n t}(1-e^{(-\sum_n k_n(t))})$$

The effective time constant for the decay of A (and effective rate constant) corresponds to the sum of all rate constants

The ratio of product concentrations depends on rate constants:

$$ \frac{[B]}{[C]} = \frac{k_b}{k_c} \text{\hspace{1 cm}independent of time!}$$

Yield of each product:

$$ \frac{[B]_{\infty}}{[B]_{\infty} + [C]_{\infty}} = \frac{k_b}{k_b + k_c} = \phi_B \text{\hspace{1cm} yield}$$

in general:

$$ \phi_j = \frac{k_j}{\sum_n k_n} $$

### _Lindemann - Hinshelwood Mechanism_

Consider collision of gas molecule $A$ with another gas molecule $M$, producing $A$ in excited state $A^*$

$$ A + M \rightarrow A* + M \hspace{1cm} \frac{dA^*}{dt} = k_a AM$$

$A*$ can lose its internal energy by collision:

$$ A^* + M \rightarrow A + M \hspace{1cm} \frac{dA^*}{dt} = -k_a^{'} A^*M$$

Alternatively, $A*$ can react to products in unimolecular reactions:

$$ A^* \rightarrow P \hspace{1cm} \frac{dA^*}{dt} = -k_b A^*$$

Reaction equation: 

$$ A + M \rightleftharpoons A A^* + M \rightarrow P + M $$

$A^*$ is short-lived (assumption) $->$ steady-state approximation

$$ \frac{dA^*}{dt} = k_a AM - k_a^{'} A^*M - k_b A^* \approx 0 $$

$$ [A^*] = \frac{k_a [A][M]}{k_b + k_a^{'}[M]} $$

Products:

$$ \frac{d[P]}{dt} = k_b [A*] = \frac{k_b k_a [A][M]}{k_b + k_a^{'}[M]} $$

If deactivation is much faster than product formation:

$$ k_a^{'} [A^*][M] >> k_b[A^*] $$

$$ \rightarrow k_a^{'} [M] >> k_b $$

"high pressure limit"

Can neglect $k_b$ in denominator for $[A^*]$ and $\frac{d[P]}{dt}$

$$ \frac{d[P]}{dt} \approx \frac{k_b k_a[A][M]}{k_a['][M]} = \frac{k_b k_a}{k_a^{'}}[A] $$

Looks like first order rate law!
