## 1.) Langevvin Model for Ion-Molecule Collisions

$$\text{Ar}^+ + \text{N}_2 \xrightarrow{k_r} \text{N}_2^+ + \text{Ar}$$

### 1a.) 

$$V_{eff}(r) = \frac{L^2}{2 \mu r^2} - \frac{\alpha e^2}{2 (4 \pi \epsilon_0)^2 r^4}$$

```julia
using Plots

gr()

b = 10e-10
ϵ_0 = 8.85e-12
α = 1.76e-30 * (4 * pi * ϵ_0)
m_Ar = 39.944
m_N2 = 28.016
μ = ((m_Ar * m_N2) / (m_Ar + m_N2)) / (6.022e+26)
k_B = 1.380649e-23
T = 300
v_rel = sqrt((8 * k_B * T) / (pi * μ))
L = μ * v_rel * b
e = 1.602176634e-19

r_range = range(1e-10, 20e-10, length=1000)     # Angstroms
r_range_plot = range(1,20, length = 1000)

V_eff_J = ((L^2) ./ (2 .* μ .* r_range.^2)) - ((α * e^2) ./ (2 .* (4 * pi * ϵ_0)^2 .* r_range.^4))
V_eff_meV = V_eff_J * 6.242e21


r_star = (e / ((4 * pi * ϵ_0) * L)) * sqrt(2 * μ * α)
V_star = ((L^2) / (2 * μ * r_star^2)) - ((α * e^2) / (2 * (4 * pi * ϵ_0)^2 * r_star^4))


plot(
    r_range_plot,
    V_eff_meV,
    title="Langevin Model",
    xlabel=raw"Distance $(r)$ $[\AA]$",
    ylabel=raw"Effective Potential $(V_{eff})$ [meV]",
    label=raw"$b = 10 \AA$",
    ylims = (-300,300),
    framestyle=:box,
    linewidth=2.0,
    size = (900, 700),
    dpi = 1000,
)

b2 = 5e-10
L_5 = μ * v_rel * b2
V_eff_J_5 = ((L_5^2) ./ (2 .* μ .* r_range.^2)) - ((α * e^2) ./ (2 .* (4 * pi * ϵ_0)^2 .* r_range.^4))
V_eff_meV_5 = V_eff_J_5 * 6.242e21

plot!(
    r_range_plot,
    V_eff_meV_5,
    label=raw"$b = 5 \AA$",
    linewidth=2.0,
    color=:pink,
)

vline!(
    [r_star] * 1e10,
    label=raw"$r^*$",
    linestyle=:dash,
    linewidth=1.5,
    color=:red,
)

hline!(
    [V_star] * 6.242e21,
    label=raw"$V^*$",
    linestyle=:dash,
    linewidth=1.5,
    color=:orange,
)
```

### 1b.)

$$E = \frac{1}{2} \mu \lang v_{rel} \rang^2$$


```julia
E = (1/2) * μ * v_rel^2 * 6.242e21
println("The energy with a speed at 300 K is: ", round(E, digits=3), " meV")
```

This is not enough energy to get over the barrier, but with $b = 5 \AA$, there is sufficient energy.

### 1c.)

$$b^* = \left[ \frac{1}{(4 \pi \epsilon_0)^2} \frac{2 \alpha e^2}{E} \right]^{1/4}$$

$$\sigma_L = \pi {b^{*}}^2 \frac{\pi e}{4 \pi \epsilon_0} \sqrt{\frac{2 \alpha}{E}}$$

$$k_L = 2 \pi \left(\frac{e}{4 \pi \epsilon_0} \right) \sqrt{\frac{\alpha}{\mu}}$$

```julia
b_star = (((1) / (4 * pi * ϵ_0)^2) * ((2 * α * e^2) / (E / 6.242e21)))^(1/4) * 1e10
println("The critical impact parameter at 300 K is: ", round(b_star, digits=3), " angstroms")

σ_L = pi * (b_star / 1e10)^2 * 1e20
println("The Langevin cross section at 300 K is: ", round(σ_L, digits=3), " angstroms^2")

k_L = 2 * pi * ((e) / (4 * pi * ϵ_0)) * sqrt(α / μ) * 1e6
println("The Langevin rate constant at 300 K is: ", round(k_L, digits=11), " cm^3 /s")
```

## 2.)

$$ V_M(r) = D[1-e^{-\alpha(r-r_e)}]^2$$

### 2a.) 

$$|\alpha \Delta r| << 1$$

$$e^{-\alpha(r-r_e)} \approx 1 - \alpha (r-r_e) + \frac{1}{2} [\alpha (r-r_e)]^2 - ...$$

$$1-e^{-\alpha(r-r_e)} \approx 1 - [1 - \alpha (r-r_e)] = \alpha (r-r_e)$$

$$V_M(r) \approx D [\alpha (r-r_e)]^2 = D\alpha^2 (r-r_e)^2$$

$$\frac{1}{2} K = D\alpha^2$$

$$K = 2D\alpha^2$$

$$\omega_0 = \sqrt{\frac{K}{\mu}} = \alpha \sqrt{\frac{2D}{\mu}}$$

```julia
using Plots

gr()

range_norm = LinRange(-2, 5, 1000)

V_M_norm = (1 .- exp.( .- range_norm)).^2
V_HO_norm = (range_norm).^2

plot(
    range_norm,
    V_M_norm,
    title="Comparison of Morse and Harmonic Oscillator Potentials",
    xlabel=raw"$\alpha(r-r_e)$",
    ylabel=raw"$V/D$",
    label="Morse Potential",
    ylims = (0,2),
    framestyle=:box,
    linewidth=2.0,
    size = (900, 700),
    dpi = 1000,
)

plot!(
    range_norm,
    V_HO_norm,
    label="Harmonic Oscillator Potential",
    linewidth=2.0,
    color=:red,
)
```

### 2b.) 

$$E_v = \hbar\omega_0 \left(v+\frac{1}{2}\right) - \frac{\hbar^2 \omega_0^2}{4D} \left(v + \frac{1}{2} \right)^2$$

$$\frac{dE_n}{dv} = \hbar \omega_0 - \frac{\hbar^2 \omega_0^2}{2D}\left(v_{max} + \frac{1}{2} \right) = 0$$

$$1 - \frac{\hbar \omega_0}{2D} \left(v_{max}+ \frac{1}{2} \right) = 0$$

$$v_{max} + \frac{1}{2} = \frac{2D}{\hbar \omega_0}$$

$$v_{max} = \frac{2D}{\hbar \omega_0} - \frac{1}{2}$$

$$\Delta v = \pm 1$$

$$v_{max + 1} = \frac{2D}{\hbar \omega_0} + \frac{1}{2}$$

For the harmonic oscillator, the energy levels are:

$$E_n = \hbar\omega_0 \left(n+\frac{1}{2}\right)$$

if $E \le D$

$$\hbar\omega_0 \left(n+\frac{1}{2}\right) \le D$$

$$n \le \frac{D}{\hbar \omega_0} - \frac{1}{2}$$

## 3.) Runge-Kutte Numerical Integration

```julia
using CSV
using DataFrames
using Plots

function kinetics_simulation()
    A_0 = 1.0         # mol/L
    B_0 = 0.0      # mol/L
    k_1 = 1         # s^-1
    k_2 = 0.1         # s^-1
    delta_t = 0.05     # s
    T = 5           # s


    t_array = Float64[]
    A_array = Float64[]
    B_array = Float64[]

    A = A_0
    B = B_0
    t = 0.0

    open("S.Arnold_PS6_Q3_Simple_output.txt", "w") do io
        write(io, "# Parameters: A_0=$A_0 mol/L, B_0=$B_0 mol/L, k_1=$k_1 s^-1, k_2=$k_2 s^-1, delta_t=$delta_t s, Simulation Time=$T s\n")
        write(io, "Time(s)\t[A](mol/L)\t[B](mol/L)\n")

        while t <= T
            push!(t_array, t)
            push!(A_array, A)
            push!(B_array, B)
            
            write(io, "$t\t$A\t$B\n")
            
            # Rate laws
            dA = (-k_1 * A + k_2 * B) * delta_t
            dB = (k_1 * A - k_2 * B) * delta_t
            
            A += dA
            B += dB
            t += delta_t
        end
    end

    df = CSV.read("S.Arnold_PS6_Q3_Simple_output.txt", DataFrame, delim='\t', comment="#")
    print(df)

    Plots.plot(
        t_array, 
        A_array,
        title="Simple First-Order Kinetics Simulation",
        label="[A]", 
        xlabel="Time [s]",
        ylabel="Concentration [mol/L]",
        color=:blue, 
        linewidth=2,
        size = (900, 700),
        dpi = 1000,
    )

    plot!(
        t_array, 
        B_array, 
        label="[B]", 
        color=:red, 
        linewidth=2
    )
end

kinetics_simulation()
```

```julia
using CSV
using DataFrames
using Plots

function kinetics_simulation()
    A_0 = 1.0         # mol/L
    B_0 = 0.0      # mol/L
    k_1 = 1         # s^-1
    k_2 = 0.1         # s^-1
    delta_t = 0.05     # s
    T = 5           # s


    t_array = Float64[]
    A_array = Float64[]
    B_array = Float64[]

    A = A_0
    B = B_0
    t = 0.0

    f_A(A, B) = -k_1 * A + k_2 * B
    f_B(A, B) =  k_1 * A - k_2 * B

    open("S.Arnold_PS6_Q3_RK4_output.txt", "w") do io
        write(io, "# Parameters: A_0=$A_0 mol/L, B_0=$B_0 mol/L, k_1=$k_1 s^-1, k_2=$k_2 s^-1, delta_t=$delta_t s, Simulation Time=$T s\n")
        write(io, "Time(s)\t[A](mol/L)\t[B](mol/L)\n")

        while t <= T
            push!(t_array, t)
            push!(A_array, A)
            push!(B_array, B)
            
            write(io, "$t\t$A\t$B\n")
            
            A_k_1 = f_A(A,B)
            B_k_1 = f_B(A,B)

            A_k_2 = f_A(A + A_k_1 * (delta_t/2), B + B_k_1 * (delta_t/2))
            B_k_2 = f_B(A + A_k_1 * (delta_t/2), B + B_k_1 * (delta_t/2))

            A_k_3 = f_A(A + A_k_2 * (delta_t/2), B + B_k_2 * (delta_t/2))
            B_k_3 = f_B(A + A_k_2 * (delta_t/2), B + B_k_2 * (delta_t/2))

            A_k_4 = f_A(A + A_k_3 * delta_t, B + B_k_3 * delta_t)
            B_k_4 = f_B(A + A_k_3 * delta_t, B + B_k_3 * delta_t)
            
            A += (delta_t/6) * (A_k_1 + 2*A_k_2 + 2*A_k_3 + A_k_4)
            B += (delta_t/6) * (B_k_1 + 2*B_k_2 + 2*B_k_3 + B_k_4)
            t += delta_t
        end
    end

    df = CSV.read("S.Arnold_PS6_Q3_RK4_output.txt", DataFrame, delim='\t', comment="#")
    print(df)

    Plots.plot(
        t_array, 
        A_array,
        title="RK4 First-Order Kinetics Simulation",
        label="[A]", 
        xlabel="Time [s]",
        ylabel="Concentration [mol/L]",
        color=:blue, 
        linewidth=2,
        size = (900, 700),
        dpi = 1000,
    )

    plot!(
        t_array, 
        B_array, 
        label="[B]", 
        color=:red, 
        linewidth=2
    )
end

kinetics_simulation()
```

For the exact solution:

$$[A](t) = [A_0] \left(1 - \frac{k_1}{k_1 + k_2} [1-e^{-(k_1 + k_2)t}] \right)$$

```julia
using CSV
using DataFrames
using Plots

function kinetics_simulation()
    A_0 = 1.0         # mol/L
    B_0 = 0.0      # mol/L
    k_1 = 1         # s^-1
    k_2 = 0.1         # s^-1
    delta_t = 0.05     # s
    T = 5           # s


    t_array = Float64[]
    dev_simple = Float64[]
    dev_rk4 = Float64[]

    A_simple, A_rk4 = A_0, A_0
    B_simple, B_rk4 = B_0, B_0
    t = 0.0

    f_A(A, B) = -k_1 * A + k_2 * B
    f_B(A, B) =  k_1 * A - k_2 * B


    open("S.Arnold_PS6_Q3_Exact_output.txt", "w") do io
        write(io, "# Parameters: A_0=$A_0 mol/L, B_0=$B_0 mol/L, k_1=$k_1 s^-1, k_2=$k_2 s^-1, delta_t=$delta_t s, Simulation Time=$T s\n")
        write(io, "Time(s)\t[A](mol/L)\n")

        while t <= T
            push!(t_array, t)

            A_exact = A_0 * (1 - (k_1 / (k_1 + k_2)) * (1 - exp(-(k_1 + k_2) *t)))
            
            write(io, "$t\t$A_exact\n")

            push!(dev_simple, A_simple - A_exact)
            push!(dev_rk4, A_rk4 - A_exact)

            dA = (-k_1 * A_simple + k_2 * B_simple) * delta_t
            dB = (k_1 * A_simple - k_2 * B_simple) * delta_t
            
            A_simple += dA
            B_simple += dB
            

            A_k_1 = f_A(A_rk4,B_rk4)
            B_k_1 = f_B(A_rk4,B_rk4)

            A_k_2 = f_A(A_rk4 + A_k_1 * (delta_t/2), B_rk4 + B_k_1 * (delta_t/2))
            B_k_2 = f_B(A_rk4 + A_k_1 * (delta_t/2), B_rk4 + B_k_1 * (delta_t/2))

            A_k_3 = f_A(A_rk4 + A_k_2 * (delta_t/2), B_rk4 + B_k_2 * (delta_t/2))
            B_k_3 = f_B(A_rk4 + A_k_2 * (delta_t/2), B_rk4 + B_k_2 * (delta_t/2))

            A_k_4 = f_A(A_rk4 + A_k_3 * delta_t, B_rk4 + B_k_3 * delta_t)
            B_k_4 = f_B(A_rk4 + A_k_3 * delta_t, B_rk4 + B_k_3 * delta_t)
            
            A_rk4 += (delta_t/6) * (A_k_1 + 2*A_k_2 + 2*A_k_3 + A_k_4)
            B_rk4 += (delta_t/6) * (B_k_1 + 2*B_k_2 + 2*B_k_3 + B_k_4)
            
            

            t += delta_t
        end
    end

    df = CSV.read("S.Arnold_PS6_Q3_Exact_output.txt", DataFrame, delim='\t', comment="#")
    print(df)

    Plots.plot(
        t_array, 
        dev_simple,
        title="Simulation Deviations (Simple vs RK4)",
        label="Simple", 
        xlabel="Time [s]",
        ylabel=raw"Deviation from $[A]_{exact}$",
        color=:blue, 
        linewidth=2,
        size = (900, 700),
        dpi = 1000,
    )

    plot!(
        t_array, 
        dev_rk4, 
        label="RK4", 
        color=:red, 
        linewidth=2
    )

end

kinetics_simulation()
```
